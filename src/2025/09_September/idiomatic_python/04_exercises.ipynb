```python
from collections import deque
```

```python
def sliding_max(nums: list, k: int) -> list:
    num_positions = deque()
    results = []
    for i, num in enumerate(nums):
        while num_positions and num_positions[0] <= i - k:
            num_positions.popleft()  # remove leftmost character if does not belong to the window

        while num_positions and nums[num_positions[0]] <= num:
            num_positions.popleft()

        num_positions.append(i)

        if i + 1 >= k:  # k iterations for the first max value
            results.append(nums[num_positions[0]])

    return results


sliding_max([1, 2, 3, 4, 5], 3)
sliding_max([2, 1, 3, 4, 1], k=2)
```

```python
import bisect


def sliding_median(nums: list, k: int) -> list:
    def _median(sorted_nums: list) -> float:
        print(f"sorted_nums: {sorted_nums}")
        nums_len = len(sorted_nums)
        mid = nums_len // 2
        if nums_len % 2 == 0:
            return (sorted_nums[mid - 1] + sorted_nums[mid]) / 2
        else:
            return sorted_nums[mid]

    sorted_window = sorted(nums[:k])
    tracker = deque(nums)
    results = [_median(sorted_window)]

    for num in nums[k:]:
        # Add the new number in a way that keeps the window sorted
        bisect.insort(sorted_window, num)
        # Remove the first character according to the initial list of numbers
        removed = tracker.popleft()
        idx = bisect.bisect_left(sorted_window, removed)  # index of the number to be removed
        sorted_window.pop(idx)

        results.append(_median(sorted_window))

    return results


print(sliding_median([2, 1, 3, 4, 1, 1, 4], k=4))
```

```python
from itertools import combinations


def sum_of_combs_of_k_match_target(nums: list, k: int, target) -> list:
    all_combs = combinations(sorted(nums), r=k)
    return list({comb for comb in all_combs if sum(comb) == target})


sum_of_combs_of_k_match_target([1, 1, 2, 2, 3, 4, 5], 3, 8)
```

```python
from collections import Counter


def k_most_frequent(nums: list, k: int) -> list:
    return list(Counter(nums).most_common(k))


k_most_frequent([2, 2, 2, 1, 1, 1], 1)
```

Task: given a list of intervals where each interval is represented as a tuple (start, end).
Your task is to merge all overlapping intervals and return a list of the merged intervals sorted by their start time.

```python
import random

intervals = [(1, 3), (2, 4), (5, 7), (6, 8)]
random.shuffle(intervals)
random.seed()


def merge_overlapping_intervals(intervals) -> list:
    if not intervals:
        return []
    sorted_intervals = sorted(intervals, key=lambda interval: interval[0])
    result = []
    current = list(sorted_intervals[0])
    for _next in sorted_intervals[1:]:
        next_start, next_end = _next
        curr_start, curr_end = current
        if curr_end >= next_start:
            current[1] = max(curr_end, next_end)
        else:
            result.append(current)
            current = list(_next)

    result.append(current)
    return result


print(merge_overlapping_intervals(intervals))
print(merge_overlapping_intervals([(1, 2), (3, 4), (5, 6)]))
print(merge_overlapping_intervals([]))
```

```python
import heapq

meetings = [(0, 30), (5, 10), (6, 15), (10, 14), (15, 31)]


def number_of_meeting_rooms(intervals) -> int:
    if not intervals:
        return 0

    sorted_meetings = sorted(intervals, key=lambda interval: interval[0])
    free_room_ends = []

    for start, end in sorted_meetings:
        if free_room_ends and start >= free_room_ends[0]:
            heapq.heappop(free_room_ends)  # remove the previous meeting which has ended

        heapq.heappush(free_room_ends, end)

    return len(free_room_ends)


print(number_of_meeting_rooms([(0, 30), (5, 10), (15, 20)]))  # 2
print(number_of_meeting_rooms([(7, 10), (2, 4)]))  # 1
print(number_of_meeting_rooms([(1, 5), (2, 6), (4, 7)]))  # 3
print(number_of_meeting_rooms([]))  # 0
```

```python
def is_operator(value):
    operators = {"+", "-", "*", "/"}
    return value in operators


def is_number(value):
    numbers = {str(num) for num in range(0, 10)}
    digits = map(str, list(value))
    return all(digit in numbers for digit in digits)


def op(value1, operator, value2):
    if operator == "+":
        return value1 + value2
    elif operator == "-":
        return value2 - value1
    elif operator == "*":
        return value1 * value2
    else:
        return value2 / value1


def solve(expression: str) -> int:
    list_of_values = expression.split()
    operators = list(filter(is_operator, list_of_values))
    numbers = list(filter(is_number, list_of_values))

    while operators:
        operator = operators.pop()
        value1, value2 = int(numbers.pop()), int(numbers.pop())
        result = op(value1, operator, value2)
        numbers.append(result)

    return int("".join(map(str, numbers)))


TEST_CASES = [
    ("+ 3 4", 7),
    ("- 3 * 4 5", -17),
    ("* 3 * 4 / 10 2", 60),
]

for test_case in TEST_CASES:
    assert solve(test_case[0]) == test_case[1]
```
